## 1. String + String--->

```python
a = "Talha"
b = "Shaikh"

print(a + b)      # TalhaShaikh
print(a+ " " +b)  # withspace
print(a == b)     # False
print(a != b)     # True
print(a > b)      # False
print(a < b)      # True
print(a * 2)      # TalhaTalha
#print(a / 2)      # TypeError: unsupported operand type(s) for /: 'str' and 'int'
print("a" in a)   # True
print("z" in a)   #False
print("z" not in a) #True
```

## 2. String + Intege --->

```python
a = "Talha"
b = 10

print(a + str(b))     # Talha10
print(a + b)          # TypeError: can only concatenate str (not "int") to str
print(a * b)          # Talha repeated 10 times
print(a == b)         # False
print(a != b)         # True
print(a >> b)         # TypeError: unsupported operand type(s) for >>: 'str' and 'int'
print(a << b)         # TypeError: unsupported operand type(s) for >>: 'str' and 'int'
print(a >= b )        # TypeError: '>=' not supported between instances of 'str' and 'int'
```

# 3. String + Float--->

```python
a = "Price = "
b = 99.99

print(a + str(b))     # Price = 99.99
#print (a + b)         # TypeError: can only concatenate str (not "float") to str
print (b - 2 )        # 97.99
print(b * 2)          # 199.9
print(b ** 2)         # 9998.0001
#print (a / 2)         # TypeError: unsupported operand type(s) for /: 'str' and 'int'
print( b / 2)         # 49.995
print ( b // 2)       # 49.0
print (b % 2 )        # 1.9899999999999949
print(a * 2)          # Price = Price =
print(a == b)         # False
print(a != b)         # True
```

## 4. String + Boolea--->

```python
a = "Value = "
b = True

print(a + str(b))     # Value = True
#print (a + b )        # TypeError: can only concatenate str (not "bool") to str
print(a * 2)          # Value = Value =
print(a * b)          # Value = 
#print(a % b)          # TypeError: not all arguments converted during string formatting
print(a == b)         # False
#print (a >> b )       # TypeError: unsupported operand type(s) for >>: 'str' and 'bool'
#print (b >= a )       # TypeError: '>=' not supported between instances of 'bool' and 'str'  
print(a != b)         # True
```

## 5. Integer + Integer--->

```python
a = 10
b = 3

print(a + b)    # 13
print(a - b)    # 7
print(a * b)    # 30
print(a / b)    # 3.333...
print(a // b)   # 3
print(a % b)    # 1
print(a ** b)   # 1000
print(a > b)    # True
print(a < b)    # False
print(a == b)   # False
print(a != b )  #True
print(b - a)    #-7
```

## 6. Integer + Float--->


```python
#True  = 1
#False = 0

a = 10
b = True

print(a + b)    # 11
print(a - b)    # 9
print(a * b)    # 10
print(a == b)   # False
print(a > b)    # True
print(a < b)    # false
print(a % b)    # 0
print(a / b)    # 10.0
print(a // b)   # 10
print(a >= b)   # True
print(a <= b)   # False
```

## 8. Float + Float--->

```python
a = 10.5
b = 2.5

print(a + b)    # 13.0
print(a - b)    # 8.0
print(a * b)    # 26.25
print(a / b)    # 4.2
print(a > b)    # True
print(a == b)   # False
print(a / b)    # 4.2
print(a // b)   # 4.0
print(a % b)    # 0.5
print(a / 8)    # 1.3125
```

## 9. Float + Boolean--->

```python
a = 10.5
b = True

print(a + b)    # 11.5
print(a - b)    # 9.5
print(b - 5)    # -4
print(a * b)    # 10.5
print(a == b)   # False
print(a > b)    # True
print(a / b)    # 10.5
print(a // b)   # 10.0
print(a % b)    # 0.5
print(a / 8)    # 1.3125
print(b / a)    # 0.09523809523809523
print(b / 2)    # 0.5
```

## 10. Boolean + Boolean--->

```python
a = True
b = False

print(a and b)  # False
print(a or b)   # True
print(not a)    # False

print(a == b)   # False
print(a != b)   # True
print(a > b)    # True
print(a >= b)   # True
print(a - b)    # 1
print(a+ 5)     # 6
print(b * 5)    # 0
```
